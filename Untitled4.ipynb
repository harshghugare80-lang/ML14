**REGRATION EQUATION**

```python
import numpy as np
X = np.array ([1,2,3,4,5],dtype=float)
Y = np.array ([2,4,5,4,5],dtype=float)
```

```python
#inital value
b0 = 1
b1 = 2
learning_rate = 0.01
epochs = 1000

n = len(X)
```

```python
# Gradient Descent
for i in range(epochs):

    # Predictions
    Y_pred = b0 + b1 * X

    # Calculate gradients
    db0 = (-2/n) * np.sum(Y - Y_pred)
    db1 = (-2/n) * np.sum(X * (Y - Y_pred))
```

```python
# Update b0 and b1
b0 = b0 - learning_rate * db0
b1 = b1 - learning_rate * db1

print("b0 =", b0)
print("b1 =", b1)

print("Regression Equation:")
print("Y =", b0, "+", b1, "X")
```
